```python
# Paso 3: Importar las funciones que necesitamos:
from imageai.Detection import ObjectDetection
```

```python
# Paso 4: Crear una variable = superpoderes de la librería
detector = ObjectDetection()
detector.setModelPath("/content/yolov3.pt")
detector.setModelTypeAsYOLOv3()
detector.loadModel()
```

```python
# Paso 5: Imprimir los objetos entrenados del modelo
print(detector.CustomObjects())
```

```python
#Paso 6: Hacer el script para la empresa
def detector_vial(image):
  objetos = detector.detector.detectObjectsFromImage(
    input_image = image, output_image_path = "/content/imagen_resultado.png",
    minimum_percentage_probability = 20
  )
  count = 0
  for objeto in objetos:
    if objeto: ["name"] == "person"
    count += 1

    if count > 0:
      print(f"Se detectaron {count} personas en el camino.")
      print("Ten cuidado y recude la velocidad")

    else:
      print("No se detecta ninguna persona en el camino")
```
